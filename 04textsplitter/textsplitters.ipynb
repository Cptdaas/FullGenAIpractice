## Text Splitters in LangChain

## Overview

In **:contentReference[oaicite:0]{index=0}**, **Text Splitters** divide large text into smaller chunks so Large Language Models (LLMs) can process them efficiently.

They are essential for:
- Handling token limits
- Building RAG (Retrieval-Augmented Generation) systems
- Improving search and embedding quality
- Preserving semantic meaning

---

## 1. Character Text Splitter

### When to use
- Simple documents
- Logs
- Plain text without structure

```python
__init__(
    chunk_size: int = 4000,
    chunk_overlap: int = 200,
    length_function: Callable[[str], int] = len,
    keep_separator: bool | Literal["start", "end"] = False,
    add_start_index: bool = False,
    strip_whitespace: bool = True,
) -> None
```

## Text Splitter Configuration Parameters

This section describes the common parameters used in text splitters in  
**:contentReference[oaicite:0]{index=0}**.

---

### Parameters

| Parameter | Description | Type | Default |
|---------|------------|------|---------|
| `chunk_size` | Maximum size of each text chunk returned by the splitter | `int` | `4000` |
| `chunk_overlap` | Number of overlapping characters between consecutive chunks | `int` | `200` |
| `length_function` | Function used to measure the length of text chunks | `Callable[[str], int]` | `len` |
| `keep_separator` | Whether to keep the separator and where to place it in each chunk (`True` = start of chunk) | `bool \| 'start' \| 'end'` | `False` |
| `add_start_index` | If `True`, includes the starting character index of each chunk in metadata | `bool` | `False` |
| `strip_whitespace` | If `True`, removes leading and trailing whitespace from each chunk | `bool` | `True` |

---

### Exceptions

| Exception | Description |
|---------|-------------|
| `ValueError` | Raised if `chunk_size` is less than or equal to `0` |
| `ValueError` | Raised if `chunk_overlap` is less than `0` |

---

### Example Usage

```python
from langchain.text_splitter import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(
    chunk_size=4000,
    chunk_overlap=200,
    length_function=len,
    keep_separator=False,
    add_start_index=False,
    strip_whitespace=True
)

chunks = splitter.split_text(long_text)


````python
from langchain_text_splitters import RecursiveCharacterTextSplitter 

text =f"""
In publishing and graphic design, Lorem ipsum is a placeholder text commonly used to demonstrate the visual form of a document or a typeface without relying on meaningful content. Lorem ipsum may be used as a placeholder before the final copy is available.
 \n It is a long established fact that a reader will be distracted by the readable content of a page when looking at its layout. \nThe point of using Lorem ipsum is that it has a more-or-less normal distribution of letters, \t as opposed to using 'Content here, content here', \n making it look like readable English.
 \n Many desktop publishing packages and web page editors now use Lorem ipsum as their default model text              
"""
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50,
    separators=["\n\n", "\n", " ", "","\t"]
)
texts = text_splitter.split_text(text)
for i, t in enumerate(texts):
    print(f"--- Chunk {i+1} ---")
    print(t)        
````

## 2. Semantic Text Splitter
When to use

- Meaning is more important than length
- Question answering
- Knowledge bases

How it works :Uses embeddings to split text when semantic meaning changes.


````python
from langchain_experimental.text_splitter import SemanticChunker
from langchain_huggingface.embeddings import HuggingFaceEmbeddings

embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
semantic_chunker = SemanticChunker(
    embeddings=embeddings,
    breakpoint_threshold_type="standard_deviation",
    breakpoint_threshold_amount=3, 
    min_chunk_size=50,
)
semantic_texts = semantic_chunker.split_text(text)
for i, t in enumerate(semantic_texts):
    print(f"--- Semantic Chunk {i+1} ---")
    print(t)    

````

````python
from langchain_text_splitters import RecursiveCharacterTextSplitter, Language
textp="""
class Student:
    def __init__(self, name, age, grade):
        self.name = name
        self.age = age
        self.grade = grade  # Grade is a float (like 8.5 or 9.2)

    def get_details(self):
        return self.name"

    def is_passing(self):
        return self.grade >= 6.0


# Example usage
student1 = Student("Aarav", 20, 8.2)
print(student1.get_details())

if student1.is_passing():
    print("The student is passing.")
else:
    print("The student is not passing.")

"""
# initialize the text splitter for Python code
code_splitter = RecursiveCharacterTextSplitter.from_language(
    chunk_size=100,
    chunk_overlap=2,
    language=Language.PYTHON)

#performing the split 
code_texts = code_splitter.split_text(textp)
for i, t in enumerate(code_texts):
    print(f"--- Code Chunk {i+1} ---")
    print(t)    
````

````python
text = """
# Project Name: Smart Student Tracker

A simple Python-based project to manage and track student data, including their grades, age, and academic status.


## Features

- Add new students with relevant info
- View student details
- Check if a student is passing
- Easily extendable class-based design


## 🛠 Tech Stack

- Python 3.10+
- No external dependencies


## Getting Started

1. Clone the repo  
   ```bash
   git clone https://github.com/your-username/student-tracker.git

"""

# Initialize the splitter
splitter = RecursiveCharacterTextSplitter.from_language(
    language=Language.MARKDOWN,
    chunk_size=200,
    chunk_overlap=0,
)

# Perform the split
chunks = splitter.split_text(text)

for i, chunk in enumerate(chunks):
    print(f"--- Markdown Chunk {i+1} ---")
    print(chunk)
````
